# Feature Engineering

Customer-level feature engineering for the retention prediction model. This starts from the cleaned transaction data produced in the data-cleaning step and builds the RFM + behavioral feature set used to train the model.

```python
import pandas as pd
```

```python
# Load the cleaned transactions produced by the data-cleaning step
df_clean = pd.read_csv("../Data/clean_transactions.csv", parse_dates=["InvoiceDate"])
df_clean.shape
```

## Define the feature and target time windows

We need 30 days of future data to determine whether a customer purchased again, so the cutoff date is moved 30 days back from the last transaction.

```python
cutoff_date = df_clean["InvoiceDate"].max() - pd.Timedelta(days=30)
cutoff_date
```

```python
# Separating the clean data into the feature period and the target period
feature_data = df_clean[df_clean["InvoiceDate"] <= cutoff_date]
target_data = df_clean[df_clean["InvoiceDate"] > cutoff_date]

feature_data.shape, target_data.shape
```

**Feature Engineering**

Since the dataset was transactional, but our model needs customer-level information.

```python
# Feature 1: Recency
last_purchase = feature_data.groupby("CustomerID")["InvoiceDate"].max()
recency = (cutoff_date - last_purchase).dt.days
recency.head()
```

```python
# Feature 2: Frequency
frequency = feature_data.groupby("CustomerID")["InvoiceNo"].nunique()
frequency.head()
```

```python
# Feature 3: Monetary
feature_data["TotalAmount"] = feature_data["Quantity"] * feature_data["UnitPrice"]

# Calculate Monetary per customer
monetary = feature_data.groupby("CustomerID")["TotalAmount"].sum()
monetary.head()
```

```python
rfm = pd.concat([recency, frequency, monetary], axis=1)
rfm.columns = ["Recency", "Frequency", "Monetary"]
rfm.head()
```

```python
# Feature 4: Total Quantity
total_quantity = feature_data.groupby("CustomerID")["Quantity"].sum()
total_quantity.head()
```

```python
# Feature 5: Unique Products
unique_products = feature_data.groupby("CustomerID")["StockCode"].nunique()
rfm["UniqueProducts"] = unique_products
rfm.head()
```

```python
# Feature 6: Average Order Value
rfm["AverageOrderValue"] = rfm["Monetary"] / rfm["Frequency"]
rfm.head()
```

```python
# Feature 7: Purchase Days
feature_data["PurchaseDate"] = feature_data["InvoiceDate"].dt.date
unique_purchase_days = feature_data.groupby("CustomerID")["PurchaseDate"].nunique()
rfm["UniquePurchaseDays"] = unique_purchase_days
rfm.head()
```

```python
# Feature 8: Customer Lifespan
first_purchase = feature_data.groupby("CustomerID")["InvoiceDate"].min()
customer_lifespan = (last_purchase - first_purchase).dt.days
rfm["CustomerLifespan"] = customer_lifespan
rfm.head()
```

```python
# Feature 9: Orders in the last 30 days of the feature window
last_30_days_start = cutoff_date - pd.Timedelta(days=30)

orders_last_30_days = (
    feature_data[feature_data["InvoiceDate"] > last_30_days_start]
    .groupby("CustomerID")["InvoiceNo"]
    .nunique()
)

rfm["OrdersLast30Days"] = orders_last_30_days
rfm["OrdersLast30Days"] = rfm["OrdersLast30Days"].fillna(0).astype(int)
rfm.head()
```

```python
# Target: did the customer purchase again in the 30-day target window?
repeat_customers = target_data["CustomerID"].unique()
rfm["Target"] = rfm.index.isin(repeat_customers).astype(int)
rfm["Target"].value_counts()
```

## Separating Features(X) and Target (Y)

```python
# Create X containing only the model input features by removing the Target column.
X = rfm.drop("Target", axis=1)
```

```python
# Create the binary target label: 1 if the customer appears in repeat_customers, otherwise 0.
Y = rfm["Target"]
```

### Splitting training and testing data

```python
import sklearn
```

```python
#  : Split the data into training and testing sets. 80% is used for training and 20% for testing; stratify keeps the target-class proportion similar in both sets.
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.20,
    random_state=42,
    stratify=Y
)
```

```python
#  Run this step as part of the analysis/modeling workflow and inspect its output before moving to the next step.
X_train.shape, X_test.shape, Y_train.shape, Y_test.shape
```

```python
#  Import StandardScaler to standardize numerical features before Logistic Regression.
#feature scaling
from sklearn.preprocessing import StandardScaler
```

```python
#  Create the scaler that will learn the mean and standard deviation of each training feature.
scaler = StandardScaler()
```

```python
#  Fit the scaler using only the training data, then transform the training features to a standardized scale.
X_train_scaled = scaler.fit_transform(X_train)
```

```python
#  Apply the already-fitted training scaler to the test data so the test set is transformed consistently without refitting.
X_test_scaled = scaler.transform(X_test)
```

```python
#  Run this step as part of the analysis/modeling workflow and inspect its output before moving to the next step.
X_train_scaled.shape, X_test_scaled.shape
```

## Model Training

```python
# Import Logistic Regression, a binary classification algorithm suitable for predicting the 0/1 repeat-customer target.
#We are starting of with Logistic Regression since it is binary classification problem.
from sklearn.linear_model import LogisticRegression
```

```python
# Create the Logistic Regression model.
model = LogisticRegression(random_state=42)
```

```python
# Train the Logistic Regression model using the scaled training features and known training labels.
model.fit(X_train_scaled, Y_train)
```

```python
# Use the trained model to predict repeat/non-repeat labels for customers in the unseen test set.
Y_pred = model.predict(X_test_scaled)
```

```python
Y_pred[:20]
```

```python
#  : Calculate classification metrics on the test set: accuracy, precision, recall, and F1 score.
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

accuracy = accuracy_score(Y_test, Y_pred)
precision = precision_score(Y_test, Y_pred)
recall = recall_score(Y_test, Y_pred)
f1 = f1_score(Y_test, Y_pred)

accuracy, precision, recall, f1
```

```python
# Build a confusion matrix to see counts of correct and incorrect positive/negative predictions.
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(Y_test, Y_pred)

cm
```

## Random Forest

```python
# Import Random Forest, a tree-based classification algorithm used to compare its performance with Logistic Regression.
from sklearn.ensemble import RandomForestClassifier
```

```python
# Create a Random Forest with 100 decision trees. random_state=42 keeps the result reproducible.
# Random Forest is trained on the original features because tree-based models do not require feature scaling.
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
```

```python
# Train the Random Forest using the same training customers and target labels used by Logistic Regression.
rf_model.fit(X_train, Y_train)
```

```python
# Generate Random Forest class predictions for the unseen test customers.
Y_pred_rf = rf_model.predict(X_test)

Y_pred_rf[:20]
```

```python
# Generate the probability that each test customer will make a repeat purchase according to the Random Forest.
Y_pred_rf_proba = rf_model.predict_proba(X_test)[:, 1]

Y_pred_rf_proba[:10]
```

```python
# Calculate Accuracy, Precision, Recall, and F1 Score for the Random Forest predictions.
rf_accuracy = accuracy_score(Y_test, Y_pred_rf)
rf_precision = precision_score(Y_test, Y_pred_rf)
rf_recall = recall_score(Y_test, Y_pred_rf)
rf_f1 = f1_score(Y_test, Y_pred_rf)

rf_accuracy, rf_precision, rf_recall, rf_f1
```

```python
# Build a confusion matrix to see the Random Forest true/false positive and negative predictions.
rf_cm = confusion_matrix(Y_test, Y_pred_rf)

rf_cm
```

```python
#Calculate which customer behavioral features contributed most to the Random Forest decisions.
feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": rf_model.feature_importances_
}).sort_values("Importance", ascending=False)

feature_importance
```

```python
#  : Compare Logistic Regression and Random Forest on the same test set using the same evaluation metrics.
comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "Accuracy": [accuracy_score(Y_test, Y_pred), rf_accuracy],
    "Precision": [precision_score(Y_test, Y_pred), rf_precision],
    "Recall": [recall_score(Y_test, Y_pred), rf_recall],
    "F1 Score": [f1_score(Y_test, Y_pred), rf_f1]
})

comparison
```
